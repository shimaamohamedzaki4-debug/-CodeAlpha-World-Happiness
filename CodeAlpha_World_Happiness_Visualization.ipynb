# World Happiness - Data Visualization

## Objective
Visualize the World Happiness dataset to identify patterns, relationships, and differences in happiness scores across countries.

## Tools
- Python
- Pandas
- Matplotlib
- Seaborn
- Google Colab

```python
import pandas as pd

url = "https://gist.githubusercontent.com/Fliolli/322bc6ec3af45a55f95eb042626a8806/raw/World_Happiness_Report_2019.csv"

df = pd.read_csv(url)

df.head()
```

## Top 10 Happiest Countries


```python
top_10 = df.sort_values("Score", ascending=False).head(10)

top_10[["Country or region", "Score"]]
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(data=top_10, x="Score", y="Country or region")

plt.title("Top 10 Happiest Countries - 2019")
plt.xlabel("Happiness Score")
plt.ylabel("Country")

plt.show()
```

## Happiness Score vs GDP per Capita


```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="GDP per capita",
    y="Score"
)

plt.title("Happiness Score vs GDP per Capita")
plt.xlabel("GDP per Capita")
plt.ylabel("Happiness Score")

plt.show()
```

## Happiness Score vs Healthy Life Expectancy

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Healthy life expectancy",
    y="Score"
)

plt.title("Happiness Score vs Healthy Life Expectancy")
plt.xlabel("Healthy Life Expectancy")
plt.ylabel("Happiness Score")

plt.show()
```

## Correlation Analysis


```python
plt.figure(figsize=(10, 7))

correlation = df.select_dtypes(include="number").corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap - World Happiness")

plt.show()
```

## Top 10 Countries by Social Support


```python
top_social = df.sort_values("Social support", ascending=False).head(10)

sns.barplot(
    data=top_social,
    x="Social support",
    y="Country or region"
)

plt.title("Top 10 Countries by Social Support")
plt.xlabel("Social Support")
plt.ylabel("Country")

plt.show()
```

```python
highest = df.loc[df["Score"].idxmax(), ["Country or region", "Score"]]
lowest = df.loc[df["Score"].idxmin(), ["Country or region", "Score"]]

print("Highest Happiness Score:")
print(highest)

print("\nLowest Happiness Score:")
print(lowest)
```
